# Crear base de datos Intact - sujetos sin amputación

In: dataset `features_mixed_global_withRest_200ms.csv`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid')
%matplotlib inline
```

## Crear dataset Sujetos NO amputados

```python
# Dolor de cabeza... cómo se unen según sujeto, estimulo, base de datos y canal
meta_cols = ['subject', 'stimulus', 'origin_db','repetition']
# Features
features = ['VAR_STD', 'Range', 'Energy', 'Peak_to_Peak', 'VAR', 'Max', 'MNF', 'SSC']
# Channel 1..8 × features
signal_cols = [f'Channel {ch}_{feat}' for ch in range(1, 9) for feat in features]
# Estímulos de interés (nos falta el reposo)
selected_stimuli = [0, 22, 25, 27, 30, 31, 37, 40]

print(f'Features       : {len(features)}')
print(f'Canales        : 8  →  columnas de señal: {len(signal_cols)}')
print(f'Estímulos selec: {selected_stimuli}')

# dataset que contiene todas las bases de datos
path_dataset = '../data/master/features_mixed_global_withRest_200ms.csv'
df_raw_intact = pd.read_csv(path_dataset)
print(f'Shape dataset: {df_raw_intact.shape}')
df_raw_intact.head(10)
```

Filtrando solo los estimulos de interés

```python
df_intact = df_raw_intact[df_raw_intact['stimulus'].isin(selected_stimuli)][meta_cols + signal_cols].copy().reset_index(drop=True)
# Solo las bases de datos 1, 2, 4 y 5 que son NO amputados
df_intact = df_intact[df_intact['origin_db'].isin(['NinaProDB1', 'NinaProDB2', 'NinaProDB4', 'NinaProDB5'])]
print(df_intact['origin_db'].unique())
# guardar database como database_intact.csv
df_intact.to_csv('../data/master/database_intact.csv', index=False)
df_intact.head(30)
```
